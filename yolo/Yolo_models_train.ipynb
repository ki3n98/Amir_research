```python
import os
import matplotlib.pyplot as plt
import pandas as pd
from ultralytics import YOLO
%matplotlib inline
```

```python
cwd = os.getcwd()
model_dirs = ['yolo9', 'yolo12', 'yolo10']

models = list()
for project in model_dirs:
    generation_dir = os.path.join(cwd, project)

    for model_path in os.listdir(generation_dir):
        if not os.path.isdir(model_path):
            model_name = model_path[:-3]
            model = YOLO(os.path.join(generation_dir, model_path))
            models.append((project, model_name, model))

df = pd.DataFrame(models, columns=["gen", "name", "model"])
df
```

```python
data_dir = os.path.join("..", "pretrain_models", "data", "mobile detection.v7i.yolov11", "data.yaml")

epochs=50
imgsz=640
batch=8
```

```python
for i, model in df.iterrows(): 
    print(f"==> Training {model['name']}")
    model["model"].train(data=data_dir,
                        epochs=epochs,
                        imgsz=imgsz,
                        batch=batch,
                        project=model["gen"],
                        name=model["name"]
    )
```

```python
yolov9e_model = df[df.name == "yolov9e"].iloc[0]
yolov9e_model.model.train(data=data_dir,
                    epochs=50,
                    imgsz=imgsz,
                    batch=4,
                    project=yolov9e_model["gen"],
                    name=yolov9e_model["name"])
```
